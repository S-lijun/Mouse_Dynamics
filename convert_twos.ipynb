```python
import os
import pandas as pd
```

```python
def convert_twos(data_dir="./", save_dir="./"):
    cols = [
        "timestamp",
        "event",
        "x",
        "y",
        "user",
        "screen_width",
        "screen_height",
    ]

    for file in os.listdir(data_dir):
        input_path = os.path.join(data_dir, file)

        # user is everything before first ".log"
        user = file.split(".log")[0]

        df = pd.read_csv(
            input_path,
            sep=";",
            header=None,
            names=cols,
        )

        ts = pd.to_datetime(
            df["timestamp"],
            format="%Y-%m-%d %H:%M:%S,%f"
        )

        df["timestamp"] = ts.astype("int64") // 1_000_000

        # keep original filename (just convert extension to .csv)
        out_dir = os.path.join(save_dir, user)
        os.makedirs(out_dir, exist_ok=True)

        out_path = os.path.join(out_dir, file + ".csv")

        df.to_csv(out_path, index=False)
```

```python
convert_twos('TWOS-dataset/mouse_ano/mouse_ano/', 'TWOS CSV/')
```
